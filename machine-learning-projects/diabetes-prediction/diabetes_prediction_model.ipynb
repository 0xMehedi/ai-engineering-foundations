```python
#-------------------Data Loading & Exploration---------------
import pandas as pd

df = pd.read_csv("diabetes.csv")
#Show first few rows
df.head()

```

```python
#Check dataset shape
df.shape
```

```python
#Check for missing values

df.isnull().sum()
```

```python
#Show basic statistics
df.describe()
```

```python
#i felt necessary
df.info()
```

```python
#---------------Data Preprocessing-------------------
from sklearn.model_selection import train_test_split
x = df.drop("Outcome",axis = 1)
y = df["Outcome"]
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2,random_state=42)

#----------
```

```python
#--------------3 Model Training------------

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier

# Logistic Regression
LogisticR = LogisticRegression(max_iter=1000)
LogisticR.fit(x_train, y_train)
y_pred_lr = LogisticR.predict(x_test)

# Random Forest
RandomF = RandomForestClassifier(random_state=42)
RandomF.fit(x_train, y_train)
y_pred_rf = RandomF.predict(x_test)

# SVM
SVM = SVC(probability=True)
SVM.fit(x_train, y_train)
y_pred_svm = SVM.predict(x_test)

# Decision Tree
DecisionT = DecisionTreeClassifier(random_state=42)
DecisionT.fit(x_train, y_train)
y_pred_dt = DecisionT.predict(x_test)

# KNN
KNN = KNeighborsClassifier()
KNN.fit(x_train, y_train)
y_pred_knn = KNN.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, roc_curve


# Logistic Regression
as_logistic = accuracy_score(y_test, y_pred_lr)
ps_logistic = precision_score(y_test, y_pred_lr)
rs_logistic = recall_score(y_test, y_pred_lr)
f1_logistic = f1_score(y_test, y_pred_lr)
roc_auc_logistic = roc_auc_score(y_test, y_pred_lr)


# Random Forest
as_rf = accuracy_score(y_test, y_pred_rf)
ps_rf = precision_score(y_test, y_pred_rf)
rs_rf = recall_score(y_test, y_pred_rf)
f1_rf = f1_score(y_test, y_pred_rf)
roc_auc_rf = roc_auc_score(y_test, y_pred_rf)


# SVM
as_svm = accuracy_score(y_test, y_pred_svm)
ps_svm = precision_score(y_test, y_pred_svm)
rs_svm = recall_score(y_test, y_pred_svm)
f1_svm = f1_score(y_test, y_pred_svm)
roc_auc_svm = roc_auc_score(y_test, y_pred_svm)


# Decision Tree
as_dt = accuracy_score(y_test, y_pred_dt)
ps_dt = precision_score(y_test, y_pred_dt)
rs_dt = recall_score(y_test, y_pred_dt)
f1_dt = f1_score(y_test, y_pred_dt)
roc_auc_dt = roc_auc_score(y_test, y_pred_dt)


# KNN
as_knn = accuracy_score(y_test, y_pred_knn)
ps_knn = precision_score(y_test, y_pred_knn)
rs_knn = recall_score(y_test, y_pred_knn)
f1_knn = f1_score(y_test, y_pred_knn)
roc_auc_knn = roc_auc_score(y_test, y_pred_knn)


#printing results

print("------ Logistic Regression ------")
print("Accuracy:", as_logistic)
print("Precision:", ps_logistic)
print("Recall:", rs_logistic)
print("F1 Score:", f1_logistic)
print("ROC-AUC:", roc_auc_logistic)


print("\n------ Random Forest ------")
print("Accuracy:", as_rf)
print("Precision:", ps_rf)
print("Recall:", rs_rf)
print("F1 Score:", f1_rf)
print("ROC-AUC:", roc_auc_rf)


print("\n------ SVM ------")
print("Accuracy:", as_svm)
print("Precision:", ps_svm)
print("Recall:", rs_svm)
print("F1 Score:", f1_svm)
print("ROC-AUC:", roc_auc_svm)


print("\n------ Decision Tree ------")
print("Accuracy:", as_dt)
print("Precision:", ps_dt)
print("Recall:", rs_dt)
print("F1 Score:", f1_dt)
print("ROC-AUC:", roc_auc_dt)


print("\n------ KNN ------")
print("Accuracy:", as_knn)
print("Precision:", ps_knn)
print("Recall:", rs_knn)
print("F1 Score:", f1_knn)
print("ROC-AUC:", roc_auc_knn)
```

```python
import matplotlib.pyplot as plt
#Ploting ROC curve

# Probability scores for ROC

y_prob_lr = LogisticR.predict_proba(x_test)[:,1]
y_prob_rf = RandomF.predict_proba(x_test)[:,1]
y_prob_svm = SVM.predict_proba(x_test)[:,1]
y_prob_dt = DecisionT.predict_proba(x_test)[:,1]
y_prob_knn = KNN.predict_proba(x_test)[:,1]

#False positive and True positives
fpr_lr, tpr_lr, _ = roc_curve(y_test, y_prob_lr)
fpr_rf, tpr_rf, _ = roc_curve(y_test, y_prob_rf)
fpr_svm, tpr_svm, _ = roc_curve(y_test, y_prob_svm)
fpr_dt, tpr_dt, _ = roc_curve(y_test, y_prob_dt)
fpr_knn, tpr_knn, _ = roc_curve(y_test, y_prob_knn)





plt.figure(figsize=(8,6))

plt.plot(fpr_lr, tpr_lr, label="Logistic Regression")
plt.plot(fpr_rf, tpr_rf, label="Random Forest")
plt.plot(fpr_svm, tpr_svm, label="SVM")
plt.plot(fpr_dt, tpr_dt, label="Decision Tree")
plt.plot(fpr_knn, tpr_knn, label="KNN")

plt.plot([0,1],[0,1],'k--')   # baseline line

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve Comparison")

plt.legend()
plt.show()
```

```python
#------------------Best Model Selection--------------

# Compare ROC-AUC scores of all models

print("ROC-AUC Score Comparison")
print("Logistic Regression:", roc_auc_logistic)
print("Random Forest:", roc_auc_rf)
print("SVM:", roc_auc_svm)
print("Decision Tree:", roc_auc_dt)
print("KNN:", roc_auc_knn)


# Store models and their scores
models = {
    "Logistic Regression": LogisticR,
    "Random Forest": RandomF,
    "SVM": SVM,
    "Decision Tree": DecisionT,
    "KNN": KNN
}

scores = {
    "Logistic Regression": roc_auc_logistic,
    "Random Forest": roc_auc_rf,
    "SVM": roc_auc_svm,
    "Decision Tree": roc_auc_dt,
    "KNN": roc_auc_knn
}


# Choose the best model
best_model_name = max(scores, key=scores.get)
best_model = models[best_model_name]

print("\nBest Model:", best_model_name)
print("Best ROC-AUC Score:", scores[best_model_name])
```

```python
# Save the best model
import joblib

joblib.dump(best_model, "diabetes_model.pkl")

print("\nBest model saved as diabetes_model.pkl")
```
