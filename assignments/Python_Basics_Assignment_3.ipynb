```python
#Question 1: Enumerate with List

fruits = ["apple", "banana", "cherry", "mango"]

for index,fruit in enumerate(fruits):
  print(f"{index} {fruit}")
```

```python
#Question 2: Zip with Two Lists

students = ["Rahim", "Karim", "Ayesha"]
scores = [85, 90, 78]

for name,score in zip(students,scores):
  print(f"{name} scored {score}")
```

```python
#Question 3: Tuple and Loop

subjects = ("Bangla", "English", "Math", "CSE", "Economics")

for subject in subjects:
  print(subject)
```

```python
#Question 4: Set Operations
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7}
#common elements
print(set1.intersection(set2))
#unique elements of set1
print(set1.difference(set2)) #ekhane union use korinai cz union diye overall unique value ber kora hoi kintu question a bola hoise sudhu set 1 er unique koita
```

```python
#Question 5: Dictionary with Enumerate
marks = {
"Math": 80,
"Physics": 75,
"Chemistry": 85
}

for index,(subject,mark) in enumerate(marks.items()):
  print(f"{index+1} {subject}: {mark}")
```
