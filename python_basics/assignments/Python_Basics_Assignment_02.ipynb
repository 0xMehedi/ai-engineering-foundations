```python
#4. Take a string from the user and count how many vowels (a, e, i, o, u) are in the string.
string = input("input a string: ")

chk = "aeiou"
count = 0
for chr in string.lower():
  if chr in chk:
    count = count+1

print(f"Numbers of vowels in {string}: {count}")

```
